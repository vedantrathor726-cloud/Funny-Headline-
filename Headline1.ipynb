# FAKE NEWS HEADLINE GENERATOR

```python
# CONCEPT USE
print("lists = to store groups of words like names, action and place")
print ("random.choice() = to pick one item randomly from a list  ")
print("print() = to show the fake headline on screen")
print("input() = to ask the user if they want another headline")
print("while loop = to repeat the process until the user says no")
print("string concatenation or s-strings = to form a sentence using selected words")
```

```python
import random

subjects = [
    "Shahrukh Khan",
    "virat kholi",
    "narendra modi",
    " yash ",
    "a group of monkey",
    "amit shah ",
    "ghe ghe ghe"

]

actions = [
    "launches",
    "cancels",
    "dance with",
    "eats",
    "declare war on ",
    "order",
    "celebrates"
]

places_or_things = [
     "at red fort",
     "in mumbai local train",
     "a plate of samosa",
     "inside the parliament",
     "at ganga ghat",
     "during IPL Match",
     "at India Gate "
]

# start the headline generation loop
while True:
  subject = random.choice(subjects)
  action = random.choice(actions)
  place_or_thing = random.choice(places_or_things)

  headline = f" BREAKING NEWS: {subject} {action} {place_or_thing}"
  print("\n" + headline)


  user_input = input("\nDo you want another headline? (yes/no): ").strip().lower()
  if user_input == "no":
    break


  #print goodbye message
  print("\nThanks for using the fake news Headline")
```
